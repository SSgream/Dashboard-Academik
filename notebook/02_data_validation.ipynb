# Relasi


## Mahasiswa Key Tabel KRS dan Mahasiswa

```python
krs = pd.read_csv("../data/fact_krs.csv")
mahasiswa = pd.read_csv("../data/dim_mahasiswa.csv")

invalid_fk = krs[
    ~krs["mahasiswa_key"].isin(mahasiswa["mahasiswa_key"])
]

print("Foreign key tidak valid:", len(invalid_fk))
```

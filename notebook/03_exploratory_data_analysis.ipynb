# Distribusi Mahasiswa

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

mahasiswa = pd.read_csv("../data/dim_mahasiswa.csv")
prodi = pd.read_csv("../data/dim_prodi.csv")

df_mahasiswa = mahasiswa.merge(
    prodi,
    on="prodi_key",
    how="left"
)

df_mahasiswa["fakultas_nama"].value_counts().plot(
    kind="bar",
    figsize=(12, 5)
)

plt.title("Student Distribution by Faculty")
plt.xlabel("Faculty")
plt.ylabel("Number of Students")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
angkatan = df_mahasiswa["angkatan"].value_counts().sort_index()

plt.figure(figsize=(10, 5))
sns.barplot(x=angkatan.index, y=angkatan.values)

plt.title("Student Distribution by Enrollment Year")
plt.xlabel("Enrollment Year")
plt.ylabel("Number of Students")
plt.tight_layout()
plt.show()
```

# Distribusi Nilai

```python
krs = pd.read_csv("../data/fact_krs.csv")

krs_approved = krs[krs["is_approved"] == True]

print("Total KRS approved:", len(krs_approved))
print("Average Score:", krs_approved["nilai_akhir"].mean())
print("Median Score:", krs_approved["nilai_akhir"].median())
print("Standard Deviation:", krs_approved["nilai_akhir"].std())
```

```python
plt.figure(figsize=(10, 5))

sns.histplot(
    data=krs_approved,
    x="nilai_akhir",
    bins=20,
    kde=True
)

plt.title("Distribution of Academic Scores")
plt.xlabel("Final Score")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

# Distribusi Kehadiran

```python
presensi = pd.read_csv("../data/fact_presensi.csv")
tipe_kehadiran = pd.read_csv("../data/dim_tipe_kehadiran.csv")

# Gabungkan data presensi dengan nama tipe kehadiran
df_presensi = presensi.merge(
    tipe_kehadiran,
    on="tipe_kehadiran_key",
    how="left"
)

# Hitung jumlah berdasarkan nama tipe kehadiran
jumlah_kehadiran = df_presensi["nama"].value_counts()

display(jumlah_kehadiran)
```
